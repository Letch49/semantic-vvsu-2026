# Часть 1. Анатомия хорошего промпта (User Prompt)

Хороший промпт содержит **6 элементов**:

| # | Элемент | Зачем |
|---|---------|-------|
| 1 | **Роль** | Задаёт «личность» модели: эксперт, наставник, рецензент… |
| 2 | **Цель** | Что должно получиться на выходе (не процесс, а результат) |
| 3 | **Задача** | Конкретные шаги, которые модель должна выполнить |
| 4 | **Лимиты / Ограничения** | Что НЕ делать, объём, формат, язык |
| 5 | **Контекст и детали** | Вводные данные: аудитория, стек, предметная область |
| 6 | **Пример** | Как должен выглядеть ответ (формат, стиль, структура) |

### Почему это работает?

LLM предсказывает следующий токен на основе **всего контекста**.
Чем точнее контекст → тем уже пространство вероятных продолжений → тем релевантнее ответ.

Представьте: вы говорите таксисту «поехали». Он не знает куда.
А если «поехали на ул. Светланская 10, через набережную, без остановок» — результат предсказуем.

---
## Пример 1: Задача по веб-программированию

### Плохой промпт (BAD)

````python
bad_web = """
Сделай мне сайт с авторизацией.
""".strip()

print(bad_web)
````

**Что не так:**
- Нет роли → модель не знает уровень ответа (джуниор? архитектор?).
- Нет стека → выберет что угодно: PHP, Django, Express…
- Нет ограничений → может выдать 500 строк или 5.
- Нет контекста → непонятно: учебный проект? MVP? продакшен?

### Хороший промпт (GOOD)

````python
good_web = """
Роль: Ты — senior fullstack-разработчик, пишущий код-ревью для джуниора.

Цель: Реализовать страницу авторизации (логин + регистрация) для учебного проекта.

Задача:
1. Опиши структуру файлов проекта.
2. Напиши backend-эндпоинты /register и /login на FastAPI.
3. Напиши frontend-форму на HTML + fetch (без фреймворков).
4. Добавь валидацию email и пароля (минимум 8 символов).
5. Объясни каждый блок кода комментарием для новичка.

Ограничения:
- Стек: Python 3.11, FastAPI, SQLite, чистый HTML/JS.
- Без Docker, без ORM — только raw SQL для понимания.
- Максимум 150 строк на файл.
- Пароль хранить как bcrypt-хэш.

Контекст:
- Проект учебный, аудитория — студенты 2 курса.
- Студенты знают основы Python, но не работали с API.

Формат ответа:
- Каждый файл — отдельный блок кода с именем файла.
- В конце — чеклист "что проверить после запуска".
""".strip()

print(good_web)
````

**Почему это лучше:**

| Элемент | Было (BAD) | Стало (GOOD) |
|---------|------------|-------------|
| Роль | — | senior fullstack, код-ревью для джуниора |
| Цель | «сайт с авторизацией» | страница логина + регистрации для учебного проекта |
| Задача | — | 5 конкретных шагов |
| Лимиты | — | стек, объём, формат хранения паролей |
| Контекст | — | аудитория, уровень знаний |
| Пример | — | формат ответа с чеклистом |

## Пример 2: Проектирование системы

### Плохой промпт (BAD)

````python
bad_system = """
Спроектируй систему для онлайн-магазина.
""".strip()

print(bad_system)
````

**Что не так:**
- «Онлайн-магазин» — это от 3 микросервисов до Amazon. Без масштаба ответ бесполезен.
- Нет целевой нагрузки, бюджета, ограничений.
- Модель выдаст «всё обо всём» на 5 страниц или слишком абстрактную схему.

### Хороший промпт (GOOD)

````python
good_system = """
Роль: Ты — системный архитектор, готовящий техническое предложение для стартапа.

Цель: Спроектировать MVP backend-а интернет-магазина одежды.

Задача:
1. Нарисуй архитектуру (перечисли компоненты и связи текстом).
2. Опиши API-контракты основных эндпоинтов (каталог, корзина, заказ).
3. Выбери базу данных и обоснуй выбор.
4. Укажи, где и как кешировать.
5. Составь план запуска MVP за 4 недели (по неделям).

Ограничения:
- Команда: 2 backend-разработчика, 1 frontend.
- Бюджет на инфру: $200/мес (облако).
- Нагрузка на MVP: до 1000 пользователей/день.
- Стек: Python, PostgreSQL, Redis, Nginx.

Контекст:
- Ассортимент: ~500 товаров, 5 категорий.
- Оплата: через внешний платёжный шлюз (интеграция по API).
- Доставка: не в скоупе MVP.

Формат ответа:
- Каждый пункт — отдельный раздел с заголовком.
- Архитектура — в виде текстовой диаграммы (ASCII или список компонентов → стрелки).
- API — в формате: METHOD /path → request body → response body.
""".strip()

print(good_system)
````

**Почему модель даст лучший ответ:**

Модель «видит» в промпте конкретные ограничения (2 разработчика, $200, 1000 юзеров).
Это сужает пространство решений: вместо абстрактной архитектуры Amazon она сгенерирует
реалистичный план для маленькой команды. Формат ответа гарантирует, что архитектура
будет структурирована так, как вам нужно.

> **Правило:** чем конкретнее ограничения и контекст, тем полезнее и проверяемее ответ.

примером и ситуацией, когда её применять.

## 2.1 Zero-Shot Prompting

**Что это:** Даём модели задачу без каких-либо примеров. Просто инструкция.

**Почему работает:**
Современные LLM обучены на огромных корпусах текста и прошли instruction tuning + RLHF.
Они уже «видели» паттерны задач (классификация, перевод, суммаризация) и могут выполнять
их без демонстрации — если задача достаточно типовая.

**Когда использовать:** Простые, типовые задачи (классификация, перевод, форматирование).

**Когда НЕ работает:** Сложные рассуждения, нестандартный формат, доменная специфика.

````python
zero_shot_example = """
Классифицируй текст как позитивный, негативный или нейтральный.

Текст: Мне кажется, этот фильм неплох.
Тональность:
""".strip()

print("ZERO-SHOT PROMPT:")
print(zero_shot_example)
print()
print("Ожидаемый ответ модели: Нейтральный")
print()
print("Обратите внимание: мы не дали НИ ОДНОГО примера.")
print("Модель уже понимает задачу классификации тональности из обучающих данных.")
````

---
## 2.2 Few-Shot Prompting

**Что это:** Даём модели **несколько примеров** (обычно 1-5) перед основным запросом.
Примеры показывают формат ввода → вывода, и модель учится «на лету» (in-context learning).

**Почему работает:**
Примеры задают модели **паттерн**: формат, стиль, уровень детализации.
Модель продолжает последовательность, следуя обнаруженному шаблону.
Даже случайные метки (!) помогают — потому что формат сам по себе несёт информацию.

**Когда использовать:**
- Нестандартный формат ответа.
- Классификация с нетипичными категориями.
- Когда zero-shot даёт нестабильный результат.

**Ограничения:** Не справляется со сложными задачами на рассуждение (арифметика, логика).

````python
few_shot_example = """
Определи тип обращения клиента.

Текст: "Заказ не пришёл, уже 10 дней жду!" → Жалоба
Текст: "Подскажите, есть ли доставка в Хабаровск?" → Вопрос
Текст: "Спасибо, всё супер, буду заказывать ещё!" → Благодарность
Текст: "Можно поменять размер в заказе #4521?" → Запрос на изменение

Текст: "Верните деньги, товар бракованный." → Жалоба
""".strip()

print("FEW-SHOT PROMPT:")
print(few_shot_example)
print()
print("Ожидаемый ответ модели: Жалоба")
print()
print("Мы дали 4 примера. Модель 'поняла':")
print("- формат ввода (текст в кавычках)")
print("- формат вывода (одно слово-категория)")
print("- набор возможных категорий")
````

## 2.3 Chain-of-Thought Prompting (CoT)

**Что это:** Просим модель рассуждать **пошагово**, показывая промежуточные шаги.

**Почему работает:**
LLM генерирует текст токен за токеном. Когда модель сразу пишет ответ,
она «сжимает» рассуждение в один шаг — и часто ошибается.
Когда мы просим расписать шаги, каждый промежуточный токен **становится контекстом**
для следующего. Модель буквально «думает на бумаге».

**Два варианта:**
1. **Few-shot CoT** — даём пример с рассуждением.
2. **Zero-shot CoT** — просто добавляем фразу: *«Давай подумаем шаг за шагом»*.

**Когда использовать:** Математика, логика, многошаговые задачи, анализ.

````python
# Без CoT — модель часто ошибается
no_cot = """
Нечётные числа в группе дают в сумме чётное число: 15, 32, 5, 13, 82, 7, 1.
Ответ:
""".strip()

print("БЕЗ CoT (модель часто ответит неверно):")
print(no_cot)
print()
print("Типичный неверный ответ: 'Да' (модель не считает, а угадывает)")
print("=" * 60)
````

````python
# Few-shot CoT — показываем КАК рассуждать
few_shot_cot = """
Нечётные числа в группе дают в сумме чётное число: 4, 8, 9, 15, 12, 2, 1.
Рассуждение: Нечётные числа: 9, 15, 1. Сумма = 9 + 15 + 1 = 25. 25 — нечётное. Ответ: Нет.

Нечётные числа в группе дают в сумме чётное число: 17, 10, 19, 4, 8, 12, 24.
Рассуждение: Нечётные числа: 17, 19. Сумма = 17 + 19 = 36. 36 — чётное. Ответ: Да.

Нечётные числа в группе дают в сумме чётное число: 15, 32, 5, 13, 82, 7, 1.
Рассуждение:
""".strip()

print("FEW-SHOT CoT (показываем шаги рассуждения в примерах):")
print(few_shot_cot)
print()
print("Ожидаемый ответ: Нечётные: 15, 5, 13, 7, 1. Сумма = 41. 41 — нечётное. Ответ: Нет.")
````

````python
# Zero-shot CoT — магическая фраза
zero_shot_cot = """
Я зашёл в магазин и купил 10 яблок.
2 отдал соседу, 2 — сантехнику.
Потом купил ещё 5 и одно съел.
Сколько яблок у меня осталось?

Давай подумаем шаг за шагом.
""".strip()

print("ZERO-SHOT CoT (просто добавляем 'Давай подумаем шаг за шагом'):")
print(zero_shot_cot)
print()
print("Без этой фразы модель часто отвечает '11' (неверно).")
print("С фразой модель расписывает: 10 - 2 - 2 = 6, 6 + 5 = 11, 11 - 1 = 10. Ответ: 10.")
````

## 2.4 Self-Consistency

**Что это:** Генерируем **несколько ответов** с рассуждениями (CoT), затем выбираем
**наиболее частый** финальный ответ (голосование большинством).

**Почему работает:**
LLM — стохастическая система. Один прогон может пойти по неверной цепочке рассуждений.
Но если из 5 прогонов 4 дают один и тот же ответ — вероятность его корректности значительно выше.
Это как спросить 5 экспертов и взять мнение большинства.

**Когда использовать:**
- Задачи с однозначным ответом (математика, логика, факты).
- Когда один запрос даёт нестабильные результаты.

**Ограничения:** Дороже (нужно несколько запросов), не подходит для творческих задач.

````python
self_consistency_example = """
Задача: Когда мне было 6 лет, моей сестре было вдвое меньше.
Сейчас мне 70. Сколько лет сестре?

--- Прогон 1 ---
Когда мне 6, сестре 6/2 = 3. Разница = 3 года. Сейчас мне 70, сестре 70 - 3 = 67.
Ответ: 67.

--- Прогон 2 ---
Мне 6, сестре вдвое меньше = 3. Разница в возрасте не меняется = 3.
70 - 3 = 67. Ответ: 67.

--- Прогон 3 ---
Мне 6, сестре 3. Сейчас мне 70, сестре 70/2 = 35. Ответ: 35.

=== Голосование ===
67 — 2 голоса, 35 — 1 голос.
Итог: 67 (правильный ответ).
""".strip()

print("SELF-CONSISTENCY (голосование из нескольких CoT-прогонов):")
print(self_consistency_example)
print()
print("Прогон 3 ошибся (поделил 70 на 2 вместо вычитания разницы).")
print("Но большинство (2 из 3) дали верный ответ 67.")
````

## 2.5 Generate Knowledge Prompting

**Что это:** Сначала просим модель **сгенерировать знания** (факты) по теме,
а затем используем эти знания как контекст для ответа на вопрос.

**Почему работает:**
LLM хранит знания «размазанными» по весам. Прямой вопрос может не активировать
нужные знания. Но если сначала попросить модель *вспомнить* факты по теме,
эти факты попадают в контекст и помогают дать точный ответ.

Аналогия: перед экзаменом вы сначала повторяете конспект, а потом отвечаете на билет.

**Когда использовать:**
- Вопросы, требующие фактических знаний.
- Задачи здравого смысла, где модель может «забыть» очевидное.

````python
# Шаг 1: Генерируем знания
gen_knowledge_step1 = """
Сгенерируй 3 ключевых факта о правилах гольфа, которые касаются подсчёта очков.
""".strip()

# Ожидаемые знания от модели:
generated_knowledge = """
1. В гольфе побеждает игрок с НАИМЕНЬШИМ количеством ударов.
2. Каждый удар считается как 1 очко. Меньше ударов = лучше.
3. «Par» — это ожидаемое количество ударов на лунку. Результат ниже par — хорошо.
""".strip()

# Шаг 2: Используем знания для ответа
gen_knowledge_step2 = f"""
Знания:
{generated_knowledge}

Вопрос: Цель гольфа — набрать больше очков, чем другие. Да или Нет?
Объясни и ответь.
""".strip()

print("ШАГ 1 — генерируем знания:")
print(gen_knowledge_step1)
print()
print("Модель сгенерировала:")
print(generated_knowledge)
print()
print("=" * 60)
print()
print("ШАГ 2 — используем знания:")
print(gen_knowledge_step2)
print()
print("Ожидаемый ответ: Нет. В гольфе побеждает тот, кто сделал МЕНЬШЕ ударов.")
print()
print("Без генерации знаний модель часто отвечает 'Да' (неверно).")
````

## 2.6 Prompt Chaining (Цепочки промптов)

**Что это:** Разбиваем сложную задачу на **последовательность простых подзадач**.
Выход одного промпта → вход следующего.

**Почему работает:**
- Каждый промпт решает одну конкретную задачу → меньше шансов на ошибку.
- Можно проверить результат каждого шага.
- Легко отладить: если ответ плохой, видно на каком этапе сломалось.

Аналогия: конвейер на заводе. Каждый рабочий делает одну операцию качественно,
вместо того чтобы один рабочий собирал весь автомобиль.

**Когда использовать:**
- Любая задача из 3+ шагов.
- Анализ документов, генерация отчётов, написание кода.
- Когда один большой промпт даёт кашу.

````python
# Плохо: один гигантский промпт
bad_chain = """
Прочитай статью, выдели главные тезисы, найди противоречия,
напиши рецензию и предложи 3 улучшения.
""".strip()

# Хорошо: цепочка промптов
chain = [
    "Промпт 1: Прочитай статью и выдели 5 основных тезисов. Формат: нумерованный список.",
    "Промпт 2: Вот тезисы: [вставить]. Есть ли между ними противоречия? Перечисли.",
    "Промпт 3: На основе тезисов и противоречий напиши рецензию на 200 слов.",
    "Промпт 4: Предложи 3 конкретных улучшения статьи с обоснованием.",
]

print("BAD — всё за раз:")
print(bad_chain)
print()
print("=" * 60)
print()
print("GOOD — цепочка промптов:")
for i, step in enumerate(chain, 1):
    print(f"  {step}")
print()
print("Каждый шаг проверяем, прежде чем идти дальше.")
print("Выход шага N → вход шага N+1.")
````

## 2.7 Tree of Thoughts (ToT)

**Что это:** Модель генерирует **несколько вариантов** на каждом шаге рассуждения,
оценивает их и продолжает только лучшие ветки. Как дерево решений.

**Почему работает:**
CoT — это одна цепочка рассуждений. Если на шаге 2 ошибка → весь ответ неверный.
ToT строит **дерево**: на каждом шаге несколько ветвей, плохие отсекаются.
Это позволяет исследовать пространство решений и находить оптимальный путь.

Аналогия: шахматы. Вы не считаете один ход вперёд — вы рассматриваете 3-4 варианта,
оцениваете каждый и выбираете лучший.

**Когда использовать:**
- Стратегическое планирование.
- Задачи с несколькими допустимыми путями решения.
- Головоломки, оптимизация.

````python
tree_of_thoughts_prompt = """
Представь, что три эксперта решают эту задачу.
Каждый эксперт записывает один шаг своего рассуждения, затем делится с группой.
После этого все переходят к следующему шагу.
Если эксперт на каком-то этапе понимает, что ошибся, он выбывает.

Задача: Как разместить 8 ферзей на шахматной доске так,
чтобы ни один не бил другого?
""".strip()

print("TREE OF THOUGHTS PROMPT:")
print(tree_of_thoughts_prompt)
print()
print("Этот промпт заставляет модель:")
print("1. Генерировать несколько вариантов параллельно (3 эксперта).")
print("2. Оценивать каждый вариант на каждом шаге.")
print("3. Отбрасывать тупиковые ветви ('эксперт выбывает').")
print("4. Продолжать только перспективные направления.")
````

## 2.8 Role Prompting (Ролевой промптинг)

**Что это:** Назначаем модели конкретную **роль** (эксперт, учитель, критик, пользователь).

**Почему работает:**
Роль активирует связанные «знания» в весах модели.
«Ты — опытный DevOps-инженер» → модель чаще использует терминологию DevOps,
предлагает решения из этой области, учитывает типичные ограничения.

**Когда использовать:** Практически всегда. Роль — первый элемент хорошего промпта.

````python
# Без роли
no_role = """
Как оптимизировать SQL-запрос?
""".strip()

# С ролью
with_role = """
Ты — DBA с 15-летним опытом работы с PostgreSQL в highload-проектах.

Студент-джуниор написал этот запрос:
SELECT * FROM orders WHERE created_at > '2024-01-01' ORDER BY total DESC;

Таблица orders: 50 млн строк, индексы — только на id.

Объясни:
1. Почему запрос медленный.
2. Какие индексы создать.
3. Как переписать запрос.
Уровень объяснения: для человека, который впервые слышит про EXPLAIN ANALYZE.
""".strip()

print("БЕЗ РОЛИ:")
print(no_role)
print()
print("С РОЛЬЮ:")
print(with_role)
````

## 2.9 Meta-Prompting (Мета-промптинг)

**Что это:** Просим модель **саму написать промпт** для решения задачи.
Или просим улучшить наш промпт.

**Почему работает:**
Модель «знает» паттерны хороших промптов из обучающих данных.
Она может структурировать задачу лучше, чем начинающий пользователь.

**Когда использовать:**
- Не знаете, как начать.
- Хотите улучшить существующий промпт.
- Нужен промпт для повторного использования (шаблон).

````python
meta_prompt = """
Я хочу, чтобы LLM помогла мне проанализировать резюме кандидатов на позицию
junior Python-разработчика.

Напиши оптимальный промпт для этой задачи.
Промпт должен включать: роль, критерии оценки, формат вывода, ограничения.
""".strip()

print("META-PROMPT (просим модель написать промпт):")
print(meta_prompt)
print()
print("Модель сгенерирует структурированный промпт, который можно")
print("сразу использовать для анализа резюме.")
````

## 2.10 ReAct (Reasoning + Acting)

**Что это:** Модель чередует **рассуждение** (Thought) и **действие** (Action).
После действия получает наблюдение (Observation) и рассуждает дальше.

**Почему работает:**
Чистое рассуждение (CoT) ограничено знаниями модели.
ReAct позволяет модели **обращаться к внешним источникам** (поиск, калькулятор, API)
и корректировать рассуждение на основе реальных данных.

**Когда использовать:**
- Задачи, требующие актуальных данных.
- Задачи с несколькими источниками информации.
- Агентные системы (AI-агенты).

````python
react_example = """
Вопрос: Какой город является столицей страны, где родился создатель Linux?

Thought 1: Нужно узнать, кто создал Linux.
Action 1: Поиск ["создатель Linux"]
Observation 1: Линус Торвальдс создал Linux в 1991 году.

Thought 2: Нужно узнать, где родился Линус Торвальдс.
Action 2: Поиск ["Линус Торвальдс место рождения"]
Observation 2: Линус Торвальдс родился в Хельсинки, Финляндия.

Thought 3: Столица Финляндии — Хельсинки. Это и есть ответ.
Action 3: Завершить ["Хельсинки"]
""".strip()

print("ReAct ПАТТЕРН (Thought → Action → Observation):")
print(react_example)
print()
print("Модель не пытается ответить сразу из памяти.")
print("Она разбивает вопрос на шаги и 'обращается' к источникам.")
````

# Часть 3. Сводная таблица техник

| Техника | Суть | Когда применять | Сложность |
|---------|------|----------------|----------|
| **Zero-shot** | Задача без примеров | Простые типовые задачи | Низкая |
| **Few-shot** | Задача с примерами | Нестандартный формат, классификация | Низкая |
| **CoT** | Пошаговое рассуждение | Математика, логика, анализ | Средняя |
| **Self-Consistency** | Несколько CoT + голосование | Задачи с точным ответом | Средняя |
| **Generate Knowledge** | Сначала факты, потом ответ | Вопросы на знания, здравый смысл | Средняя |
| **Prompt Chaining** | Цепочка подзадач | Многошаговые задачи | Средняя |
| **Tree of Thoughts** | Дерево вариантов с отсечением | Стратегия, планирование, головоломки | Высокая |
| **Role Prompting** | Назначение роли | Практически всегда | Низкая |
| **Meta-Prompting** | Модель пишет промпт | Начало работы, оптимизация | Низкая |
| **ReAct** | Рассуждение + действия | Агенты, поиск информации | Высокая |

# Часть 4. Общие правила промптинга

### 1. Будьте конкретны, а не абстрактны
- BAD: «Напиши хороший код»
- GOOD: «Напиши функцию на Python, которая принимает список чисел и возвращает среднее. Добавь docstring и обработку пустого списка»

### 2. Указывайте формат ответа
- Markdown, JSON, таблица, нумерованный список — модель подстроится.
- Без указания формата ответ будет непредсказуемым.

### 3. Разделяйте данные и инструкции
- Используйте разделители: `---`, `###`, `"""`, XML-теги `<context>...</context>`.
- Это помогает модели отличить «что анализировать» от «что делать».

### 4. Говорите, что делать, а не что НЕ делать
- BAD: «Не используй сложные слова»
- GOOD: «Объясняй простыми словами, как для ученика 9 класса»

### 5. Используйте итеративный подход
- Первый промпт редко идеален. Уточняйте, добавляйте контекст, меняйте формат.
- Промптинг — это диалог, а не одноразовый запрос.

### 6. Задавайте ограничения
- Длина ответа, количество пунктов, стек технологий, целевая аудитория.
- Ограничения сужают пространство ответов и повышают релевантность.

### 7. Проверяйте ответ модели
- LLM может уверенно выдавать неверную информацию (галлюцинации).
- Всегда верифицируйте факты, код и расчёты.
- Просите модель саму указать, в чём она не уверена.

### 8. Контекст > длина промпта
- Длинный промпт без контекста хуже, чем короткий с контекстом.
- Дайте модели то, что ей нужно для ответа, и уберите лишнее.
